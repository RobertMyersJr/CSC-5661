```python
import numpy as np
import random
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification
```

```python
class Contextual_Bandit_generator:

    #@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@
    def __init__(self, config):
        self.state = {'X':[],'article':[],'reward':[]}
        self.population = [[],[]]
        self.population[0], self.population[1] = make_classification(
            n_samples=config['n_samples'],
            n_features=5,
            n_informative=5,
            n_classes=5,
            n_redundant=0,
            n_repeated=0,
            flip_y=0.2,
            shift = [10,4,0,0,0],
            scale = [7,1,1,1,1],
            random_state=8
        )
        self.population[0] = self.feature_engineer(self.population[0])
        self.current_context = {'X':[],'best_article':[]}
        
    #@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@
    def feature_engineer(self, Data):
        Data[:,0] = (Data[:,0]>0)*1
        Data[:,1] = (Data[:,1]>0)*1
        Data[:,3][(Data[:,3]<0)] = 0
        Data[:,3][(Data[:,3]>5)] = 5
        Data[:,3] = np.round(Data[:,3])
        Data[:,4] = np.round(Data[:,4])
        return Data

    #@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@
    def step(self):
        sample = np.random.choice(len(self.population[1]))
        x = self.population[0][sample]
        ba = self.population[1][sample]
        if len(self.state['X']) == 0:
            print(f"Setting x: {x}")
            self.state['X'] = x
        else:
            self.state['X'] = np.vstack((self.state['X'],x))
        self.current_context['X'] = x
        self.current_context['best_article'] = ba

    #@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@@
    def reward(self,a):        
        if self.current_context['best_article'] == a:
            response = np.random.binomial(n = 1, p = 0.9,size = 1)[0]
        else:
            response = np.random.binomial(n = 1, p = 0.05,size = 1)[0]
        if len(self.state['X']) == 0:      
            self.state['article'] = np.array(a)
            self.state['reward'] = np.array(response)
        else:
            self.state['article'] = np.append(self.state['article'],a)
            self.state['reward'] = np.append(self.state['reward'],response)
        
```

```python
import random
import numpy as np
from sklearn.ensemble import RandomForestClassifier

turns = 1000
T = 100
prob_e = 0.5
b = 50

# Actions must match the class labels generated by make_classification (0 to 4)
actions = [0, 1, 2, 3, 4]

f = {act: RandomForestClassifier() for act in actions}
env_config = {'n_samples': 100000}
env = Contextual_Bandit_generator(env_config)

X_data = {a: [] for a in actions}
y_data = {a: [] for a in actions}

for t in range(turns):
    print(t)
    # 1. Advance the environment step
    env.step()
    
    # Get current feature vector from current_context
    X_curr = env.current_context['X']

    # 2. Choose action (Explore vs Exploit)
    if t <= T:
        a = random.choice(actions)
    else:
        prob = random.random()
        if prob < prob_e:
            a = random.choice(actions)
        else:
            scores = [
                f[act].predict_proba(X_curr.reshape(1, -1))[0][1] 
                if hasattr(f[act], "classes_") and 1 in f[act].classes_ 
                else 0.0
                for act in actions
            ]
            a = actions[np.argmax(scores)]

    # 3. Apply reward and store training data
    env.reward(a)
    
    X_data[a].append(X_curr)
    y_data[a].append(env.state["reward"][t])

    # 4. Periodically refit models after burn-in
    if t > T and t % b == 0:
        for act in actions:
            if len(X_data[act]) > 0 and len(np.unique(y_data[act])) > 1:
                f[act].fit(np.array(X_data[act]), y_data[act])

print("Average reward: " + str(np.average(env.state['reward'])))
```
